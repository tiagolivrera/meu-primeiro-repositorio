## **PEP - 8**


O PEP - 8 consiste nas Propostas de Aprimoramento do Python (Python Enhancement Proposal). Ele é um guia para se ter um bom código para a linguagem de programação python, que pode promover um melhor uso da linguagem e suas funcionalidades, bem como podem ajudar em questões como arquitetura, ambiente ou processos de sua aplicação.

> Na sequência, serão mostradas algumas dessas convenções:


> * Comentários: É aconselhável inserir em seu código comentários que sinalizem as decisões que estão sendo tomadas, de prefêrencia que sejam claros e objetivos, para adicionar esses comentários basta utilizar a cerquilha ou hashtag ( # ), 6 apóstrofos ( ''' comentário ''' ) ou 6 aspas ( """ comentário """ );


````python
# Essa é uma forma de adicionar um comentário
#bom dia
#hoje
````

````python
'''Essa é outra forma de adicionar comentários

hoje é um dia bonito


'''
````

````python
"""Aqui é mais uma forma de adicionar comentários"""
````


> * Indentação: a cada nível de indentação é aconselhável utilizar 4 espaços, as linhas de código devem ser alinhadas verticalmente a partir da linha ímplicita do Python, juntando-se entre parênteses, chaves ou colchetes;

````python
#Essa é a forma aconselhável:
x = (1 + 2 + 3 + 4
  + 5 + 6 + 7)

#Não é aconselhável dessa forma:
x = (1 + 2
+ 3 + 4 + 5
+ 6 + 7)
````

> * Espaços em operações: é aconselhável sempre utilizar espaços antes e depois da colocação dos operadores (não importando a natureza);

````python
#Essa é a forma aconselhável:
2 + 2
4 - 1
5 - 3

#Não é aconselhável dessa forma:
2+2
4-1
````

> * Quebra de linha em operações: mesmo que as operações estejam entre parênteses é aconselhável colocar o operador após a quebra de linha para se ter uma melhor legibilidade da operação desejada;

````python
#Essa é a forma aconselhável:
x = (1
    + 2
    - 3)

#Não é aconselhável dessa forma:
x = (1 +
     2 -
     3)
````

> * Nomes de variáveis: é aconselhável evitar nomes de variáveis com letras únicas do tipo I, l, O, pois a chance de ser confundida com um número é grande. Nós recomendamos a utilização palavras associadas à descrição da funcionalidade, com todas as letras minúsculas e separadas com underlines ( _ ), isso serve também para funções. Esta convenção é conhecida como _snake_case_ (Ex.: calcula_integral).

````python
#Essa é a forma aconselhável:
soma_dois_elementos = x + y

sub_dois_numeros = x - y

#Não é aconselhável dessa forma:
ca-dfk89a0 = x + y
````

> * Comprimento máximo de linha: é aconselhável limitar todas as linhas a um máximo de 79 caracteres. Limitar a largura da janela do editor torna possível ter vários arquivos abertos lado a lado;

> * Linhas em branco: são aconselháveis quando se quer ter uma melhor organização do seu bloco de códigos;

> Essas orientações mostradas são algumas de uma grande quantidade que pode ser vista na própria [documentação do PEP-8](https://www.python.org/dev/peps/pep-0008/).

# **Operadores Comparativos**

Operador relacional ou comparativo é todo operador que obtém a relação do membro a esquerda com o membro a sua direita.

## **Igual ( == )**

Se o valor das variáveis propostas a serem comparadas forem iguais, teremos uma condição verdadeira (True). Sendo assim a condição ( x == y ) é True.

````python
x = 4
y = 4

print(x == y)
````

Caso contrário, a condição será False:

````python
x = 5
y = 6

print(x == y)
````

**OBS.:** Nota-se que quando utilizado apenas ' = ' a linguagem de programação entende como atribuição para alguma variável, já quando utilizado ' == ' é entendido como o operador comparativo.

## **Maior que ( > )**

Caso o valor da variável localizada a esquerda do símbolo ' > ' for maior que a localizada a direita, a condição é verdadeira (True). Sendo assim, a condição ( x > y ) é True.

````python
x = 27
y = 12

print(x > y)
````

Caso contrário, a condição será False:

````python
x = 53
y = 62

print(x > y)
````

or.JPG

# **Estruturas Condicionais**

As estuturas condicionais têm como seu maior objetivo dar ao programa o poder de decisão sobre quais comandos executar, essas estruturas tem alguns elementos, como o if ( se ), elif ( abreviação de else if, que significa outro se) e o else ( outro ).

## **if**

No caso de um if simples, o programa irá seguir apenas a condição determinada, tendo como sua estrutura a seguinte:



```
 if condição :
    comandos
```



Tomando como exemplo um alarme de segurança de uma residência:

````python
AlarmePorta = False
AlarmeJanela = False

if AlarmePorta == True or AlarmeJanela == True:
  print('ALERTA! OS ALARMES DA PORTA E DA JANELA FORAM DISPARADOS!')
  print('Continua')
````

Pode-se também haver vários if's em uma estrutura, sendo representada dessa forma:



```
if ( condição1 ) :
    comandos

if ( condição2 ) :
    comandos

if ( condiçãoN ) :
    comandos
```

Ainda utlizando o exemplo anterior:

````python
AlarmePorta = True
AlarmeJanela = True

if (AlarmePorta == True and AlarmeJanela == False):
  print('ALERTA! O ALARME DA PORTA FOI DISPARADO!')

if (AlarmeJanela == True and AlarmePorta == False):
  print('ALERTA! O ALARME DA JANELA FOI DISPARADO!')

if (AlarmePorta == True and AlarmeJanela == True):
  print('ALERTA! OS ALARMES DA PORTA E DA JANELA FORAM DISPARADOS!')
````

## **else**

```
if ( condição ) :
  comandos

else:
  comandos
```

Utilizando o mesmo exemplo dos alarmes:


````python
AlarmePorta = True

if (AlarmePorta == True):
  print('ALERTA! O ALARME DA PORTA FOI DISPARADO!')

else:
  print('A PORTA ESTÁ SEGURA!')
````

# **Definição de Função**

Funções são blocos de código que realizam determinadas tarefas que normalmente precisam ser executadas diversas vezes dentro de uma aplicação. Quando surge essa necessidade, para que várias instruções não precisem ser repetidas, elas são agrupadas em uma função, à qual é dado um nome e que poderá ser chamada/executada em diferentes partes do programa.

A sintaxe de uma função é definida por três partes: nome, parâmetros e corpo, o qual agrupa uma sequência de linhas que representa algum comportamento.



```
def nome(parâmetros):
  corpo
```



Tomando como exemplo um empresa com um grande quadro de funcionários, é preciso calcular o salário para cada um, mesmo tendo cargos e cargas horárias diferentes. Sendo assim, para não precisar fazer um cálculo para cada um individualmente, é mais prático definir uma função para nos retornar a quantia do salário.

````python
def calcular_salario(quantidade_horas, valor_hora):
                           # O nome dado a função é calcular_salario e os parâmetros são quantidade_horas e valor_hora.
  horas = quantidade_horas # Essa variávei representa as horas trabalhadas no mês.
  taxa  = valor_hora       # Essa variável representa o valor da hora trabalhada.

  if horas <= 160:         # 160 representa o número de horas trabalhadas no mês, caso o funcionário trabalhe 8 horas por dia útil.

    salario = horas*taxa

  else:                    # Caso o funcionário resolva fazer hora extra, o cálculo do salário deveserá ser diferente.

    hora_extra  = horas - 160
    salario = 160 * taxa + ( hora_extra * ( 1.5 * taxa ) )

  return salario       # O return representa qual a informação a função irá nos fornecer assim que acionada.
````

Então agora verificaremos o salário de um funcionário que trabalhe as 8 horas por dia útil (ou seja, 160 horas no mês), com o valor da hora de R$ 7,575 :

````python
x = calcular_salario(160, 7.575)
x
````

Ou seja, o funcionário 1 receberá R$ 1212 (o valor do salário mínimo). Caso o funcionário 2 tenha o mesmo custo de hora trabalhada porém resolveu trabalhar 9 horas por dia útil (180 horas no mês):

````python
calcular_salario(180, 7.575)
````

````python
def media(nota1, nota2, nota3):
  '''
  Essa função calcula a média de 3 notas e retorna o valor da média.
  '''
  # Cálculo
  valor_media = (nota1 + nota2 + nota3) / 3
  return valor_media
````
